```python
import json
import math
from pathlib import Path

import numpy as np
import pandas as pd


# --------------------------------------------------
# Helper functions
# --------------------------------------------------

def parse_json_list(value):
    """
    Convert a JSON string, list, tuple or NumPy array
    into a normal Python list.
    """

    if isinstance(value, str):
        return json.loads(value)

    if isinstance(value, np.ndarray):
        return value.tolist()

    if isinstance(value, tuple):
        return list(value)

    if isinstance(value, list):
        return value

    raise TypeError(
        f"Unsupported JSON-list type: {type(value)}"
    )


def format_cell(value):
    """
    Create a stable textual representation of a table cell.
    """

    if value is None:
        return "missing"

    if isinstance(value, float):
        if math.isnan(value):
            return "missing"

        # Avoid unnecessary trailing zeros
        return f"{value:.10g}"

    return str(value).strip()


def serialize_table(columns_json, rows_json):
    """
    Serialize a structured table while explicitly preserving
    column names and row-level category-value relationships.
    """

    columns = parse_json_list(columns_json)
    rows = parse_json_list(rows_json)

    column_names = [
        format_cell(column)
        for column in columns
    ]

    text_parts = [
        "table columns: "
        + " | ".join(column_names)
    ]

    for row_number, row in enumerate(rows, start=1):
        row = parse_json_list(row)

        cell_pairs = []

        for column_index, column_name in enumerate(
            column_names
        ):
            if column_index < len(row):
                cell_value = format_cell(
                    row[column_index]
                )
            else:
                cell_value = "missing"

            cell_pairs.append(
                f"{column_name} = {cell_value}"
            )

        text_parts.append(
            f"row {row_number}: "
            + " ; ".join(cell_pairs)
        )

    return ". ".join(text_parts) + "."


# --------------------------------------------------
# Confirm that table IDs are unique
# --------------------------------------------------

duplicate_table_ids = (
    tables_df["data_id"]
    .duplicated()
    .sum()
)

print("Duplicate table data IDs:", duplicate_table_ids)

assert duplicate_table_ids == 0, (
    "Table data_id must be unique for oracle mapping."
)


# --------------------------------------------------
# Create canonical table representations
# --------------------------------------------------

canonical_tables_df = tables_df.copy()

canonical_tables_df["canonical_table_text"] = (
    canonical_tables_df.apply(
        lambda row: serialize_table(
            row["columns_json"],
            row["rows_json"]
        ),
        axis=1
    )
)

canonical_tables_df = canonical_tables_df.rename(
    columns={
        "variant_type": "table_variant_type",
        "table_text": "raw_table_text"
    }
)


# --------------------------------------------------
# Map every chart to its underlying table
# --------------------------------------------------

table_lookup_columns = [
    "data_id",
    "source_data_id",
    "split",
    "table_variant_type",
    "raw_table_text",
    "canonical_table_text",
    "table_hash"
]

oracle_charts_df = charts_df.merge(
    canonical_tables_df[table_lookup_columns],
    on=[
        "data_id",
        "source_data_id",
        "split"
    ],
    how="left",
    validate="many_to_one",
    indicator=True
)


# --------------------------------------------------
# Validate the chart-to-table mapping
# --------------------------------------------------

print("\nChart-to-table merge status:")

print(
    oracle_charts_df["_merge"]
    .value_counts(dropna=False)
)

unmatched_charts = oracle_charts_df[
    oracle_charts_df["_merge"] != "both"
]

print("\nUnmatched charts:", len(unmatched_charts))

assert len(unmatched_charts) == 0, (
    "Some charts could not be mapped to tables."
)

oracle_charts_df = (
    oracle_charts_df
    .drop(columns="_merge")
)


# --------------------------------------------------
# Inspect mapping by variant
# --------------------------------------------------

variant_mapping = pd.crosstab(
    oracle_charts_df["variant_type"],
    oracle_charts_df["table_variant_type"],
    margins=True
)

print("\nChart variant → table variant mapping")

display(variant_mapping)


# --------------------------------------------------
# Display sample original and hard-negative tables
# --------------------------------------------------

sample_columns = [
    "chart_id",
    "data_id",
    "variant_type",
    "table_variant_type",
    "canonical_table_text"
]

print("\nOriginal chart-table mapping")

display(
    oracle_charts_df[
        oracle_charts_df["table_variant_type"]
        == "original"
    ][sample_columns].head(2)
)

print("\nHard-negative chart-table mapping")

display(
    oracle_charts_df[
        oracle_charts_df["table_variant_type"]
        == "hard_negative"
    ][sample_columns].head(2)
)


# --------------------------------------------------
# Save derived oracle inputs
# --------------------------------------------------

ORACLE_DATA_DIR = (
    PROJECT_ROOT
    / "data"
    / "derived"
    / "oracle_e5"
)

ORACLE_DATA_DIR.mkdir(
    parents=True,
    exist_ok=True
)

canonical_tables_path = (
    ORACLE_DATA_DIR
    / "canonical_tables.parquet"
)

oracle_charts_path = (
    ORACLE_DATA_DIR
    / "oracle_chart_table_mapping.parquet"
)

canonical_tables_df.to_parquet(
    canonical_tables_path,
    index=False
)

oracle_charts_df.to_parquet(
    oracle_charts_path,
    index=False
)

print("\nSaved files")
print(
    "Canonical tables:",
    canonical_tables_path,
    canonical_tables_path.exists()
)

print(
    "Oracle chart mapping:",
    oracle_charts_path,
    oracle_charts_path.exists()
)
```

```python
from sentence_transformers import SentenceTransformer
from tqdm.auto import tqdm

import numpy as np
import pandas as pd
import torch


# --------------------------------------------------
# Model configuration
# --------------------------------------------------

MODEL_NAME = "intfloat/e5-base-v2"

MODEL_CACHE_DIR = (
    PROJECT_ROOT
    / "artifacts"
    / "model_cache"
)

MODEL_CACHE_DIR.mkdir(
    parents=True,
    exist_ok=True
)

device = (
    "cuda"
    if torch.cuda.is_available()
    else "cpu"
)

print("Device:", device)

if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))


# --------------------------------------------------
# Load E5
# --------------------------------------------------

e5_model = SentenceTransformer(
    MODEL_NAME,
    device=device,
    cache_folder=str(MODEL_CACHE_DIR)
)

# Explicitly use the model's maximum supported length
e5_model.max_seq_length = 512

e5_tokenizer = e5_model.tokenizer

print("Model:", MODEL_NAME)
print(
    "Embedding dimension:",
    e5_model.get_sentence_embedding_dimension()
)
print(
    "Maximum sequence length:",
    e5_model.max_seq_length
)


# --------------------------------------------------
# Prepare passage text
# --------------------------------------------------

# E5 expects the "passage:" prefix for candidate
# documents such as charts and tables.
canonical_tables_df["e5_passage_text"] = (
    "passage: "
    + canonical_tables_df[
        "canonical_table_text"
    ].astype(str)
)


# --------------------------------------------------
# Count tokens without truncating
# --------------------------------------------------

def count_e5_tokens(text):
    tokens = e5_tokenizer.tokenize(text)

    # Add [CLS] and [SEP]
    return len(tokens) + 2


token_counts = []

for text in tqdm(
    canonical_tables_df["e5_passage_text"],
    desc="Counting E5 table tokens"
):
    token_counts.append(
        count_e5_tokens(text)
    )

canonical_tables_df["e5_token_count"] = (
    token_counts
)

canonical_tables_df["requires_chunking"] = (
    canonical_tables_df["e5_token_count"]
    > e5_model.max_seq_length
)


# --------------------------------------------------
# Token statistics
# --------------------------------------------------

number_truncated = int(
    canonical_tables_df[
        "requires_chunking"
    ].sum()
)

truncation_rate = (
    number_truncated
    / len(canonical_tables_df)
)

print("\nE5 token statistics")
print(
    "Mean tokens:",
    round(
        canonical_tables_df[
            "e5_token_count"
        ].mean(),
        2
    )
)

print(
    "Median tokens:",
    round(
        canonical_tables_df[
            "e5_token_count"
        ].median(),
        2
    )
)

print(
    "95th percentile:",
    round(
        np.percentile(
            canonical_tables_df[
                "e5_token_count"
            ],
            95
        ),
        2
    )
)

print(
    "Maximum tokens:",
    canonical_tables_df[
        "e5_token_count"
    ].max()
)

print(
    "Tables requiring chunking:",
    number_truncated
)

print(
    "Chunking rate:",
    round(truncation_rate, 4)
)


# --------------------------------------------------
# Breakdown by table variant
# --------------------------------------------------

token_summary = (
    canonical_tables_df
    .groupby("table_variant_type")
    .agg(
        number_of_tables=(
            "data_id",
            "count"
        ),
        mean_tokens=(
            "e5_token_count",
            "mean"
        ),
        maximum_tokens=(
            "e5_token_count",
            "max"
        ),
        tables_requiring_chunking=(
            "requires_chunking",
            "sum"
        )
    )
)

display(
    token_summary.round(2)
)


# --------------------------------------------------
# Inspect the longest tables
# --------------------------------------------------

display_columns = [
    "data_id",
    "split",
    "table_variant_type",
    "e5_token_count",
    "canonical_table_text"
]

print("\nFive longest serialized tables")

display(
    canonical_tables_df[
        display_columns
    ]
    .sort_values(
        "e5_token_count",
        ascending=False
    )
    .head(5)
)
```

```python
import json

import numpy as np
import pandas as pd
import torch


# --------------------------------------------------
# Prepare table metadata in embedding order
# --------------------------------------------------

canonical_tables_df = (
    canonical_tables_df
    .reset_index(drop=True)
    .copy()
)

canonical_tables_df["embedding_index"] = np.arange(
    len(canonical_tables_df)
)

table_passages = (
    canonical_tables_df[
        "e5_passage_text"
    ]
    .astype(str)
    .tolist()
)


# --------------------------------------------------
# Prepare query metadata
# --------------------------------------------------

query_metadata_df = (
    query_ground_truth_df
    .drop_duplicates(subset="query_id")
    .reset_index(drop=True)
    .copy()
)

assert len(query_metadata_df) == 986

query_metadata_df["embedding_index"] = np.arange(
    len(query_metadata_df)
)

# E5 requires the query prefix for retrieval queries
query_metadata_df["e5_query_text"] = (
    "query: "
    + query_metadata_df["question"].astype(str)
)

query_texts = (
    query_metadata_df[
        "e5_query_text"
    ]
    .tolist()
)


# --------------------------------------------------
# Encode unique tables
# --------------------------------------------------

e5_model.eval()

with torch.inference_mode():
    table_embeddings = e5_model.encode(
        table_passages,
        batch_size=64,
        show_progress_bar=True,
        convert_to_numpy=True,
        normalize_embeddings=True
    )

table_embeddings = table_embeddings.astype(
    np.float32
)

print(
    "Table embeddings:",
    table_embeddings.shape
)


# --------------------------------------------------
# Encode queries
# --------------------------------------------------

with torch.inference_mode():
    query_embeddings = e5_model.encode(
        query_texts,
        batch_size=128,
        show_progress_bar=True,
        convert_to_numpy=True,
        normalize_embeddings=True
    )

query_embeddings = query_embeddings.astype(
    np.float32
)

print(
    "Query embeddings:",
    query_embeddings.shape
)


# --------------------------------------------------
# Map each chart to its table embedding
# --------------------------------------------------

table_index_by_data_id = dict(
    zip(
        canonical_tables_df["data_id"],
        canonical_tables_df["embedding_index"]
    )
)

oracle_charts_df[
    "table_embedding_index"
] = oracle_charts_df["data_id"].map(
    table_index_by_data_id
)

missing_embedding_mappings = (
    oracle_charts_df[
        "table_embedding_index"
    ]
    .isna()
    .sum()
)

print(
    "Charts missing table embeddings:",
    missing_embedding_mappings
)

assert missing_embedding_mappings == 0

oracle_charts_df = (
    oracle_charts_df
    .reset_index(drop=True)
    .copy()
)

oracle_charts_df["embedding_index"] = np.arange(
    len(oracle_charts_df)
)

chart_table_indices = (
    oracle_charts_df[
        "table_embedding_index"
    ]
    .astype(int)
    .to_numpy()
)

oracle_chart_embeddings = (
    table_embeddings[
        chart_table_indices
    ]
    .copy()
)

print(
    "Oracle chart embeddings:",
    oracle_chart_embeddings.shape
)


# --------------------------------------------------
# Validate embedding norms
# --------------------------------------------------

print("\nMean embedding norms")

print(
    "Tables:",
    round(
        float(
            np.linalg.norm(
                table_embeddings,
                axis=1
            ).mean()
        ),
        6
    )
)

print(
    "Charts:",
    round(
        float(
            np.linalg.norm(
                oracle_chart_embeddings,
                axis=1
            ).mean()
        ),
        6
    )
)

print(
    "Queries:",
    round(
        float(
            np.linalg.norm(
                query_embeddings,
                axis=1
            ).mean()
        ),
        6
    )
)


# --------------------------------------------------
# Original-vs-shuffled similarity diagnostic
# --------------------------------------------------

original_tables = (
    canonical_tables_df[
        canonical_tables_df[
            "table_variant_type"
        ] == "original"
    ][
        [
            "source_data_id",
            "embedding_index"
        ]
    ]
    .rename(
        columns={
            "embedding_index":
            "original_embedding_index"
        }
    )
)

shuffled_tables = (
    canonical_tables_df[
        canonical_tables_df[
            "table_variant_type"
        ] == "hard_negative"
    ][
        [
            "source_data_id",
            "embedding_index"
        ]
    ]
    .rename(
        columns={
            "embedding_index":
            "shuffled_embedding_index"
        }
    )
)

counterfactual_pairs_df = original_tables.merge(
    shuffled_tables,
    on="source_data_id",
    how="inner",
    validate="one_to_one"
)

original_vectors = table_embeddings[
    counterfactual_pairs_df[
        "original_embedding_index"
    ].to_numpy()
]

shuffled_vectors = table_embeddings[
    counterfactual_pairs_df[
        "shuffled_embedding_index"
    ].to_numpy()
]

counterfactual_similarities = np.sum(
    original_vectors * shuffled_vectors,
    axis=1
)

counterfactual_pairs_df[
    "original_shuffled_similarity"
] = counterfactual_similarities

print("\nOriginal ↔ shuffled table similarity")

print(
    "Number of pairs:",
    len(counterfactual_pairs_df)
)

print(
    "Mean:",
    round(
        float(
            counterfactual_similarities.mean()
        ),
        6
    )
)

print(
    "Median:",
    round(
        float(
            np.median(
                counterfactual_similarities
            )
        ),
        6
    )
)

print(
    "Minimum:",
    round(
        float(
            counterfactual_similarities.min()
        ),
        6
    )
)

print(
    "Maximum:",
    round(
        float(
            counterfactual_similarities.max()
        ),
        6
    )
)


# --------------------------------------------------
# Save embeddings and metadata
# --------------------------------------------------

np.save(
    EMBEDDING_DIR
    / "table_embeddings.npy",
    table_embeddings
)

np.save(
    EMBEDDING_DIR
    / "oracle_chart_embeddings.npy",
    oracle_chart_embeddings
)

np.save(
    EMBEDDING_DIR
    / "query_embeddings.npy",
    query_embeddings
)

canonical_tables_df.to_parquet(
    EMBEDDING_DIR
    / "table_metadata.parquet",
    index=False
)

oracle_charts_df.to_parquet(
    EMBEDDING_DIR
    / "oracle_chart_metadata.parquet",
    index=False
)

query_metadata_df.to_parquet(
    EMBEDDING_DIR
    / "query_metadata.parquet",
    index=False
)

counterfactual_pairs_df.to_parquet(
    EMBEDDING_DIR
    / "counterfactual_table_similarities.parquet",
    index=False
)


# --------------------------------------------------
# Save model configuration
# --------------------------------------------------

embedding_config = {
    "model_name": MODEL_NAME,
    "embedding_dimension": int(
        table_embeddings.shape[1]
    ),
    "maximum_sequence_length": int(
        e5_model.max_seq_length
    ),
    "table_prefix": "passage: ",
    "query_prefix": "query: ",
    "number_of_tables": int(
        len(canonical_tables_df)
    ),
    "number_of_charts": int(
        len(oracle_charts_df)
    ),
    "number_of_queries": int(
        len(query_metadata_df)
    ),
    "tables_requiring_chunking": int(
        canonical_tables_df[
            "requires_chunking"
        ].sum()
    )
}

with open(
    EMBEDDING_DIR
    / "embedding_config.json",
    "w",
    encoding="utf-8"
) as file:
    json.dump(
        embedding_config,
        file,
        indent=2
    )


# --------------------------------------------------
# Verify saved files
# --------------------------------------------------

saved_files = [
    "table_embeddings.npy",
    "oracle_chart_embeddings.npy",
    "query_embeddings.npy",
    "table_metadata.parquet",
    "oracle_chart_metadata.parquet",
    "query_metadata.parquet",
    "counterfactual_table_similarities.parquet",
    "embedding_config.json"
]

print("\nSaved files")

for file_name in saved_files:
    file_path = EMBEDDING_DIR / file_name

    print(
        file_name,
        ":",
        file_path.exists()
    )
```

```python
import json

import numpy as np
import pandas as pd


# --------------------------------------------------
# Embedding lookup dictionaries
# --------------------------------------------------

chart_index_by_id = dict(
    zip(
        oracle_charts_df["chart_id"],
        oracle_charts_df["embedding_index"]
    )
)

table_index_by_id = dict(
    zip(
        canonical_tables_df["data_id"],
        canonical_tables_df["embedding_index"]
    )
)


# --------------------------------------------------
# Helper function
# --------------------------------------------------

def retrieve_indices(ids, lookup, label):
    indices = pd.Series(ids).map(lookup)

    missing_count = int(
        indices.isna().sum()
    )

    print(
        f"{label} missing embeddings:",
        missing_count
    )

    assert missing_count == 0, (
        f"Some {label} IDs do not have embeddings."
    )

    return indices.astype(int).to_numpy()


def summarize_pairwise(
    dataframe,
    negative_label
):
    positive_similarity = dataframe[
        "positive_similarity"
    ]

    negative_similarity = dataframe[
        "negative_similarity"
    ]

    gap = dataframe["similarity_gap"]

    return {
        "number_of_pairs": int(len(dataframe)),
        "mean_positive_similarity": float(
            positive_similarity.mean()
        ),
        "mean_negative_similarity": float(
            negative_similarity.mean()
        ),
        "mean_similarity_gap": float(
            gap.mean()
        ),
        "median_similarity_gap": float(
            gap.median()
        ),
        "positive_wins_rate": float(
            (gap > 0).mean()
        ),
        f"{negative_label}_wins_rate": float(
            (gap < 0).mean()
        ),
        "exact_tie_rate": float(
            np.isclose(
                positive_similarity,
                negative_similarity,
                atol=1e-7,
                rtol=0
            ).mean()
        ),
        "gap_above_0_001_rate": float(
            (gap >= 0.001).mean()
        ),
        "gap_above_0_01_rate": float(
            (gap >= 0.01).mean()
        )
    }


# ==================================================
# 1. Oracle chart → chart diagnostic
# ==================================================

oracle_chart_chart_df = (
    retrieval_pairs_df[
        retrieval_pairs_df["task"]
        == "chart_chart"
    ]
    .reset_index(drop=True)
    .copy()
)

print(
    "Chart→chart triplets:",
    len(oracle_chart_chart_df)
)

anchor_chart_indices = retrieve_indices(
    oracle_chart_chart_df["anchor_id"],
    chart_index_by_id,
    "chart anchors"
)

positive_chart_indices = retrieve_indices(
    oracle_chart_chart_df["positive_id"],
    chart_index_by_id,
    "positive charts"
)

negative_chart_indices = retrieve_indices(
    oracle_chart_chart_df["negative_id"],
    chart_index_by_id,
    "negative charts"
)

anchor_chart_vectors = oracle_chart_embeddings[
    anchor_chart_indices
]

positive_chart_vectors = oracle_chart_embeddings[
    positive_chart_indices
]

negative_chart_vectors = oracle_chart_embeddings[
    negative_chart_indices
]

oracle_chart_chart_df[
    "positive_similarity"
] = np.sum(
    anchor_chart_vectors
    * positive_chart_vectors,
    axis=1
)

oracle_chart_chart_df[
    "negative_similarity"
] = np.sum(
    anchor_chart_vectors
    * negative_chart_vectors,
    axis=1
)

oracle_chart_chart_df[
    "similarity_gap"
] = (
    oracle_chart_chart_df[
        "positive_similarity"
    ]
    - oracle_chart_chart_df[
        "negative_similarity"
    ]
)

chart_metadata_lookup = (
    oracle_charts_df
    .set_index("chart_id")
)

oracle_chart_chart_df[
    "anchor_chart_type"
] = (
    oracle_chart_chart_df["anchor_id"]
    .map(
        chart_metadata_lookup[
            "chart_type"
        ]
    )
)

oracle_chart_chart_df[
    "positive_chart_type"
] = (
    oracle_chart_chart_df["positive_id"]
    .map(
        chart_metadata_lookup[
            "chart_type"
        ]
    )
)

oracle_chart_chart_df[
    "chart_type_transition"
] = (
    oracle_chart_chart_df[
        "anchor_chart_type"
    ]
    + "_to_"
    + oracle_chart_chart_df[
        "positive_chart_type"
    ]
)

chart_chart_summary_rows = [
    {
        "group": "overall",
        **summarize_pairwise(
            oracle_chart_chart_df,
            "style_negative"
        )
    }
]

for transition, group_df in (
    oracle_chart_chart_df.groupby(
        "chart_type_transition",
        sort=False
    )
):
    chart_chart_summary_rows.append(
        {
            "group": transition,
            **summarize_pairwise(
                group_df,
                "style_negative"
            )
        }
    )

oracle_chart_chart_metrics_df = (
    pd.DataFrame(chart_chart_summary_rows)
    .set_index("group")
)

print("\nOracle E5 chart→chart diagnostic")

display(
    oracle_chart_chart_metrics_df.round(6)
)


# ==================================================
# 2. Oracle chart → table diagnostic
# ==================================================

oracle_chart_table_df = (
    retrieval_pairs_df[
        retrieval_pairs_df["task"]
        == "chart_table"
    ]
    .reset_index(drop=True)
    .copy()
)

print(
    "\nChart→table pairs:",
    len(oracle_chart_table_df)
)

chart_query_indices = retrieve_indices(
    oracle_chart_table_df["anchor_id"],
    chart_index_by_id,
    "chart queries"
)

positive_table_indices = retrieve_indices(
    oracle_chart_table_df["positive_id"],
    table_index_by_id,
    "positive tables"
)

negative_table_indices = retrieve_indices(
    oracle_chart_table_df["negative_id"],
    table_index_by_id,
    "negative tables"
)

chart_query_vectors = oracle_chart_embeddings[
    chart_query_indices
]

positive_table_vectors = table_embeddings[
    positive_table_indices
]

negative_table_vectors = table_embeddings[
    negative_table_indices
]

oracle_chart_table_df[
    "positive_similarity"
] = np.sum(
    chart_query_vectors
    * positive_table_vectors,
    axis=1
)

oracle_chart_table_df[
    "negative_similarity"
] = np.sum(
    chart_query_vectors
    * negative_table_vectors,
    axis=1
)

oracle_chart_table_df[
    "similarity_gap"
] = (
    oracle_chart_table_df[
        "positive_similarity"
    ]
    - oracle_chart_table_df[
        "negative_similarity"
    ]
)

oracle_chart_table_df[
    "chart_type"
] = (
    oracle_chart_table_df["anchor_id"]
    .map(
        chart_metadata_lookup[
            "chart_type"
        ]
    )
)

chart_table_summary_rows = [
    {
        "group": "overall",
        **summarize_pairwise(
            oracle_chart_table_df,
            "counterfactual"
        )
    }
]

for chart_type, group_df in (
    oracle_chart_table_df.groupby(
        "chart_type",
        sort=False
    )
):
    chart_table_summary_rows.append(
        {
            "group": chart_type,
            **summarize_pairwise(
                group_df,
                "counterfactual"
            )
        }
    )

oracle_chart_table_metrics_df = (
    pd.DataFrame(chart_table_summary_rows)
    .set_index("group")
)

print("\nOracle E5 chart→table diagnostic")

display(
    oracle_chart_table_metrics_df.round(6)
)


# --------------------------------------------------
# Save diagnostic results
# --------------------------------------------------

chart_chart_results_path = (
    METRICS_DIR
    / "oracle_e5_chart_chart_pairwise.parquet"
)

chart_table_results_path = (
    METRICS_DIR
    / "oracle_e5_chart_table_pairwise.parquet"
)

chart_chart_metrics_path = (
    METRICS_DIR
    / "oracle_e5_chart_chart_pairwise.json"
)

chart_table_metrics_path = (
    METRICS_DIR
    / "oracle_e5_chart_table_pairwise.json"
)

oracle_chart_chart_df.to_parquet(
    chart_chart_results_path,
    index=False
)

oracle_chart_table_df.to_parquet(
    chart_table_results_path,
    index=False
)

with open(
    chart_chart_metrics_path,
    "w",
    encoding="utf-8"
) as file:
    json.dump(
        oracle_chart_chart_metrics_df.to_dict(
            orient="index"
        ),
        file,
        indent=2
    )

with open(
    chart_table_metrics_path,
    "w",
    encoding="utf-8"
) as file:
    json.dump(
        oracle_chart_table_metrics_df.to_dict(
            orient="index"
        ),
        file,
        indent=2
    )

print("\nSaved files")
print(
    chart_chart_results_path.name,
    ":",
    chart_chart_results_path.exists()
)
print(
    chart_table_results_path.name,
    ":",
    chart_table_results_path.exists()
)
print(
    chart_chart_metrics_path.name,
    ":",
    chart_chart_metrics_path.exists()
)
print(
    chart_table_metrics_path.name,
    ":",
    chart_table_metrics_path.exists()
)
```

```python
import json

import numpy as np
import pandas as pd


# --------------------------------------------------
# Helper functions
# --------------------------------------------------

def parse_id_list(value):
    if isinstance(value, str):
        return json.loads(value)

    if isinstance(value, np.ndarray):
        return value.tolist()

    if isinstance(value, list):
        return value

    if isinstance(value, tuple):
        return list(value)

    return []


def calculate_ndcg_at_k(
    ranked_ids,
    positive_ids,
    k=10
):
    positive_ids = set(positive_ids)

    top_k_ids = ranked_ids[:k]

    relevance = np.array(
        [
            1.0 if chart_id in positive_ids else 0.0
            for chart_id in top_k_ids
        ],
        dtype=np.float32
    )

    discounts = 1.0 / np.log2(
        np.arange(2, len(relevance) + 2)
    )

    dcg = float(
        np.sum(relevance * discounts)
    )

    ideal_positive_count = min(
        len(positive_ids),
        k
    )

    if ideal_positive_count == 0:
        return 0.0

    idcg = float(
        np.sum(
            discounts[:ideal_positive_count]
        )
    )

    return dcg / idcg


def summarize_query_chart(dataframe):
    return {
        "number_of_queries": int(len(dataframe)),
        "recall_at_1": float(
            dataframe["recall_at_1"].mean()
        ),
        "recall_at_5": float(
            dataframe["recall_at_5"].mean()
        ),
        "recall_at_10": float(
            dataframe["recall_at_10"].mean()
        ),
        "mrr": float(
            dataframe["reciprocal_rank"].mean()
        ),
        "ndcg_at_10": float(
            dataframe["ndcg_at_10"].mean()
        ),
        "median_first_positive_rank": float(
            dataframe[
                "first_positive_rank"
            ].median()
        ),
        "mean_positive_similarity": float(
            dataframe[
                "mean_positive_similarity"
            ].mean()
        ),
        "mean_counterfactual_similarity": float(
            dataframe[
                "mean_counterfactual_similarity"
            ].mean()
        ),
        "mean_query_data_gap": float(
            dataframe[
                "query_data_gap"
            ].mean()
        ),
        "counterfactual_outrank_rate": float(
            dataframe[
                "counterfactual_outranks"
            ].mean()
        )
    }


# --------------------------------------------------
# Query and chart lookup tables
# --------------------------------------------------

query_index_by_id = dict(
    zip(
        query_metadata_df["query_id"],
        query_metadata_df["embedding_index"]
    )
)

chart_index_by_id = dict(
    zip(
        oracle_charts_df["chart_id"],
        oracle_charts_df["embedding_index"]
    )
)

query_chart_result_rows = []


# --------------------------------------------------
# Evaluate each split separately
# --------------------------------------------------

for split_name in ["train", "val", "test"]:

    split_charts_df = (
        oracle_charts_df[
            oracle_charts_df["split"]
            == split_name
        ]
        .reset_index(drop=True)
        .copy()
    )

    split_queries_df = (
        query_metadata_df[
            query_metadata_df["split"]
            == split_name
        ]
        .reset_index(drop=True)
        .copy()
    )

    candidate_chart_ids = (
        split_charts_df["chart_id"]
        .astype(str)
        .to_numpy()
    )

    candidate_embedding_indices = (
        split_charts_df["embedding_index"]
        .astype(int)
        .to_numpy()
    )

    candidate_embeddings = (
        oracle_chart_embeddings[
            candidate_embedding_indices
        ]
    )

    candidate_id_set = set(
        candidate_chart_ids.tolist()
    )

    candidate_local_index = {
        chart_id: index
        for index, chart_id
        in enumerate(candidate_chart_ids)
    }

    print(
        f"{split_name}:",
        len(split_queries_df),
        "queries,",
        len(candidate_chart_ids),
        "candidate charts"
    )

    for _, query_row in split_queries_df.iterrows():

        query_id = query_row["query_id"]

        query_embedding_index = (
            query_index_by_id[query_id]
        )

        query_vector = query_embeddings[
            query_embedding_index
        ]

        similarities = (
            candidate_embeddings
            @ query_vector
        )

        ranking_order = np.argsort(
            -similarities,
            kind="stable"
        )

        ranked_chart_ids = (
            candidate_chart_ids[
                ranking_order
            ]
        )

        positive_ids = [
            chart_id
            for chart_id in parse_id_list(
                query_row[
                    "positive_chart_ids_json"
                ]
            )
            if chart_id in candidate_id_set
        ]

        counterfactual_ids = [
            chart_id
            for chart_id in parse_id_list(
                query_row[
                    "counterfactual_chart_ids_json"
                ]
            )
            if chart_id in candidate_id_set
        ]

        assert len(positive_ids) > 0, (
            f"No positive charts for {query_id}"
        )

        positive_id_set = set(positive_ids)

        positive_ranks = [
            rank_position
            for rank_position, chart_id
            in enumerate(
                ranked_chart_ids,
                start=1
            )
            if chart_id in positive_id_set
        ]

        first_positive_rank = min(
            positive_ranks
        )

        positive_similarities = np.array(
            [
                similarities[
                    candidate_local_index[
                        chart_id
                    ]
                ]
                for chart_id in positive_ids
            ],
            dtype=np.float32
        )

        counterfactual_similarities = np.array(
            [
                similarities[
                    candidate_local_index[
                        chart_id
                    ]
                ]
                for chart_id
                in counterfactual_ids
            ],
            dtype=np.float32
        )

        mean_positive_similarity = float(
            positive_similarities.mean()
        )

        best_positive_similarity = float(
            positive_similarities.max()
        )

        if len(counterfactual_similarities) > 0:
            mean_counterfactual_similarity = float(
                counterfactual_similarities.mean()
            )

            best_counterfactual_similarity = float(
                counterfactual_similarities.max()
            )

            query_data_gap = (
                mean_positive_similarity
                - mean_counterfactual_similarity
            )

            counterfactual_outranks = int(
                best_counterfactual_similarity
                > best_positive_similarity
            )
        else:
            mean_counterfactual_similarity = np.nan
            best_counterfactual_similarity = np.nan
            query_data_gap = np.nan
            counterfactual_outranks = 0

        query_chart_result_rows.append(
            {
                "query_id": query_id,
                "split": split_name,
                "question": query_row["question"],
                "answer": query_row["answer"],
                "question_source": (
                    query_row["question_source"]
                ),
                "candidate_pool_size": int(
                    len(candidate_chart_ids)
                ),
                "number_of_positives": int(
                    len(positive_ids)
                ),
                "number_of_counterfactuals": int(
                    len(counterfactual_ids)
                ),
                "first_positive_rank": int(
                    first_positive_rank
                ),
                "recall_at_1": int(
                    first_positive_rank <= 1
                ),
                "recall_at_5": int(
                    first_positive_rank <= 5
                ),
                "recall_at_10": int(
                    first_positive_rank <= 10
                ),
                "reciprocal_rank": float(
                    1.0 / first_positive_rank
                ),
                "ndcg_at_10": calculate_ndcg_at_k(
                    ranked_chart_ids,
                    positive_ids,
                    k=10
                ),
                "mean_positive_similarity": (
                    mean_positive_similarity
                ),
                "mean_counterfactual_similarity": (
                    mean_counterfactual_similarity
                ),
                "best_positive_similarity": (
                    best_positive_similarity
                ),
                "best_counterfactual_similarity": (
                    best_counterfactual_similarity
                ),
                "query_data_gap": float(
                    query_data_gap
                ),
                "counterfactual_outranks": int(
                    counterfactual_outranks
                )
            }
        )


# --------------------------------------------------
# Build result dataframe
# --------------------------------------------------

oracle_query_chart_results_df = pd.DataFrame(
    query_chart_result_rows
)

assert len(oracle_query_chart_results_df) == 986


# --------------------------------------------------
# Summarize overall and by split
# --------------------------------------------------

metric_rows = [
    {
        "group": "overall",
        **summarize_query_chart(
            oracle_query_chart_results_df
        )
    }
]

for split_name, split_df in (
    oracle_query_chart_results_df.groupby(
        "split",
        sort=False
    )
):
    metric_rows.append(
        {
            "group": split_name,
            **summarize_query_chart(split_df)
        }
    )

oracle_query_chart_metrics_df = (
    pd.DataFrame(metric_rows)
    .set_index("group")
)

print("\nOracle E5 query→chart results")

display(
    oracle_query_chart_metrics_df.round(6)
)


# --------------------------------------------------
# Save results
# --------------------------------------------------

results_path = (
    METRICS_DIR
    / "oracle_e5_query_chart_results.parquet"
)

metrics_path = (
    METRICS_DIR
    / "oracle_e5_query_chart_metrics.json"
)

oracle_query_chart_results_df.to_parquet(
    results_path,
    index=False
)

with open(
    metrics_path,
    "w",
    encoding="utf-8"
) as file:
    json.dump(
        oracle_query_chart_metrics_df.to_dict(
            orient="index"
        ),
        file,
        indent=2
    )

print("\nSaved:")
print(results_path)
print(metrics_path)
```

```python
import json

import numpy as np
import pandas as pd


# --------------------------------------------------
# Summary function
# --------------------------------------------------

def summarize_oracle_chart_table(dataframe):
    return {
        "number_of_charts": int(len(dataframe)),
        "recall_at_1": float(
            dataframe["recall_at_1"].mean()
        ),
        "recall_at_5": float(
            dataframe["recall_at_5"].mean()
        ),
        "recall_at_10": float(
            dataframe["recall_at_10"].mean()
        ),
        "mrr": float(
            dataframe["reciprocal_rank"].mean()
        ),
        "ndcg_at_10": float(
            dataframe["ndcg_at_10"].mean()
        ),
        "median_positive_rank": float(
            dataframe["positive_rank"].median()
        ),
        "mean_positive_similarity": float(
            dataframe[
                "positive_similarity"
            ].mean()
        ),
        "mean_negative_similarity": float(
            dataframe[
                "negative_similarity"
            ].mean()
        ),
        "mean_table_data_gap": float(
            dataframe["similarity_gap"].mean()
        ),
        "data_over_counterfactual_accuracy": float(
            (
                dataframe["similarity_gap"] > 0
            ).mean()
        ),
        "counterfactual_outrank_rate": float(
            (
                dataframe["similarity_gap"] < 0
            ).mean()
        ),
        "exact_tie_rate": float(
            np.isclose(
                dataframe["positive_similarity"],
                dataframe["negative_similarity"],
                atol=1e-7,
                rtol=0
            ).mean()
        )
    }


# --------------------------------------------------
# Evaluate chart→table retrieval by split
# --------------------------------------------------

chart_table_retrieval_rows = []

for split_name in ["train", "val", "test"]:

    split_tables_df = (
        canonical_tables_df[
            canonical_tables_df["split"]
            == split_name
        ]
        .reset_index(drop=True)
        .copy()
    )

    split_chart_queries_df = (
        oracle_chart_table_df[
            oracle_chart_table_df["split"]
            == split_name
        ]
        .reset_index(drop=True)
        .copy()
    )

    candidate_table_ids = (
        split_tables_df["data_id"]
        .astype(str)
        .to_numpy()
    )

    candidate_embedding_indices = (
        split_tables_df["embedding_index"]
        .astype(int)
        .to_numpy()
    )

    candidate_table_embeddings = (
        table_embeddings[
            candidate_embedding_indices
        ]
    )

    candidate_table_set = set(
        candidate_table_ids.tolist()
    )

    candidate_local_index = {
        table_id: index
        for index, table_id
        in enumerate(candidate_table_ids)
    }

    print(
        f"{split_name}:",
        len(split_chart_queries_df),
        "chart queries,",
        len(candidate_table_ids),
        "candidate tables"
    )

    for _, pair_row in (
        split_chart_queries_df.iterrows()
    ):

        chart_id = pair_row["anchor_id"]
        positive_table_id = pair_row["positive_id"]
        negative_table_id = pair_row["negative_id"]

        assert positive_table_id in candidate_table_set
        assert negative_table_id in candidate_table_set

        chart_embedding_index = (
            chart_index_by_id[chart_id]
        )

        chart_vector = oracle_chart_embeddings[
            chart_embedding_index
        ]

        similarities = (
            candidate_table_embeddings
            @ chart_vector
        )

        ranking_order = np.argsort(
            -similarities,
            kind="stable"
        )

        ranked_table_ids = (
            candidate_table_ids[
                ranking_order
            ]
        )

        positive_rank = int(
            np.where(
                ranked_table_ids
                == positive_table_id
            )[0][0]
            + 1
        )

        positive_local_index = (
            candidate_local_index[
                positive_table_id
            ]
        )

        negative_local_index = (
            candidate_local_index[
                negative_table_id
            ]
        )

        positive_similarity = float(
            similarities[
                positive_local_index
            ]
        )

        negative_similarity = float(
            similarities[
                negative_local_index
            ]
        )

        similarity_gap = (
            positive_similarity
            - negative_similarity
        )

        if positive_rank <= 10:
            ndcg_at_10 = float(
                1.0
                / np.log2(
                    positive_rank + 1
                )
            )
        else:
            ndcg_at_10 = 0.0

        chart_table_retrieval_rows.append(
            {
                "pair_id": pair_row["pair_id"],
                "chart_id": chart_id,
                "split": split_name,
                "chart_type": pair_row["chart_type"],
                "positive_table_id": (
                    positive_table_id
                ),
                "negative_table_id": (
                    negative_table_id
                ),
                "candidate_pool_size": int(
                    len(candidate_table_ids)
                ),
                "positive_rank": positive_rank,
                "recall_at_1": int(
                    positive_rank <= 1
                ),
                "recall_at_5": int(
                    positive_rank <= 5
                ),
                "recall_at_10": int(
                    positive_rank <= 10
                ),
                "reciprocal_rank": float(
                    1.0 / positive_rank
                ),
                "ndcg_at_10": ndcg_at_10,
                "positive_similarity": (
                    positive_similarity
                ),
                "negative_similarity": (
                    negative_similarity
                ),
                "similarity_gap": float(
                    similarity_gap
                ),
                "counterfactual_outranks": int(
                    negative_similarity
                    > positive_similarity
                )
            }
        )


# --------------------------------------------------
# Build results dataframe
# --------------------------------------------------

oracle_chart_table_retrieval_df = pd.DataFrame(
    chart_table_retrieval_rows
)

assert len(oracle_chart_table_retrieval_df) == 2402


# --------------------------------------------------
# Summarize overall, split and chart type
# --------------------------------------------------

metric_rows = [
    {
        "group": "overall",
        **summarize_oracle_chart_table(
            oracle_chart_table_retrieval_df
        )
    }
]

for split_name, split_df in (
    oracle_chart_table_retrieval_df.groupby(
        "split",
        sort=False
    )
):
    metric_rows.append(
        {
            "group": split_name,
            **summarize_oracle_chart_table(
                split_df
            )
        }
    )

for chart_type, chart_type_df in (
    oracle_chart_table_retrieval_df.groupby(
        "chart_type",
        sort=False
    )
):
    metric_rows.append(
        {
            "group": chart_type,
            **summarize_oracle_chart_table(
                chart_type_df
            )
        }
    )

oracle_chart_table_retrieval_metrics_df = (
    pd.DataFrame(metric_rows)
    .set_index("group")
)

print("\nOracle E5 chart→table retrieval")

display(
    oracle_chart_table_retrieval_metrics_df.round(6)
)


# --------------------------------------------------
# Save results
# --------------------------------------------------

results_path = (
    METRICS_DIR
    / "oracle_e5_chart_table_results.parquet"
)

metrics_path = (
    METRICS_DIR
    / "oracle_e5_chart_table_metrics.json"
)

oracle_chart_table_retrieval_df.to_parquet(
    results_path,
    index=False
)

with open(
    metrics_path,
    "w",
    encoding="utf-8"
) as file:
    json.dump(
        oracle_chart_table_retrieval_metrics_df.to_dict(
            orient="index"
        ),
        file,
        indent=2
    )

print("\nSaved:")
print(results_path)
print(metrics_path)
```
